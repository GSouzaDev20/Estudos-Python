```python

print (f"==Calculadora==\n")

while True:

  print ("\n1 - Soma")
  print ("2 - Subtração")
  print ("3 - Multiplicação")
  print ("4 - Divisão")
  print ("0 - Sair")
  Operador = int(input(f"\nSelecione qual operador deseja utilizar!"))

  if Operador == 0:
    break
  Numero1 = int(input("\nDigite o primeiro número: "))
  Numero2 = int(input("\nDigite o segundo número: "))

  if Operador == 1:
    print(f"\nA soma dos dois números é {Numero1 + Numero2}")

  elif Operador == 2:
    print(f"\nA subtração dos dois números é {Numero1 - Numero2}")

  elif Operador == 3:
    print(f"\nA multiplicação dos dois números é {Numero1 * Numero2} ")

  elif Operador == 4:
    print(f"\nA divisão dos dois números é {Numero1 / Numero2}")

  else:
    print(f"Operador invalido")

print(f"\nSistema encerrado")

```
